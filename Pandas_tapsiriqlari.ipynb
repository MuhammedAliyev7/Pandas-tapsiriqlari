## Tapşırıq 1

```python
import pandas as pd

books = pd.DataFrame({
    "Title": ["Xəzinə", "Səfər", "Dünya", "Gələcək", "Dəniz", "Ulduzlar"],
    "Author": ["A. Əliyev", "B. Məmmədov", "C. Həsənli", "D. Quliyev", "E. İsmayılov", "F. Cəfərov"],
    "PublicationYear": [2010, 2018, 2020, 2014, 2022, 2017],
    "Price": [12.5, 18.0, 20.0, 9.5, 15.0, 17.5]
})

print(books)
print("Orta qiymət:", books["Price"].mean())
print(books[books["PublicationYear"] > 2015])
print(books.sort_values(by="Price", ascending=True))
```

## Tapşırıq 2

```python
import pandas as pd
from io import StringIO

csv_data = """OrderID,Customer,Date,Amount
1001,Aysel,2026-10-01,120.50
1002,Murad,2026-10-02,75.00
1003,Aysel,2026-10-03,210.00
1004,Kamran,2026-10-04,45.50
1005,Nigar,2026-10-05,160.00
1006,Murad,2026-10-06,95.00
1007,Kamran,2026-10-07,310.00
1008,Aysel,2026-10-08,80.00
1009,Nigar,2026-10-09,130.00
1010,Murad,2026-10-10,220.00
1011,Kamran,2026-10-11,60.00
1012,Nigar,2026-10-12,185.00
"""

with open("orders.csv", "w", encoding="utf-8") as file:
    file.write(csv_data)

orders = pd.read_csv(StringIO(csv_data))
print(orders.head(10))
print(orders.groupby("Customer").size())
print("Maksimum məbləğ:", orders["Amount"].max())
print("Minimum məbləğ:", orders["Amount"].min())
print("Ümumi məbləğ:", orders["Amount"].sum())
```

## Tapşırıq 3

```python
import pandas as pd

food = pd.DataFrame({
    "Product": ["Düyü", "Toyuq", "Yumurta", "Çörək", "Badam", "Qatıq", "Somon", "Kartof", "Pendir", "Mərci"],
    "Category": ["Taxıl", "Ət", "Süd məhsulu", "Taxıl", "Çərəz", "Süd məhsulu", "Balıq", "Tərəvəz", "Süd məhsulu", "Paxlalı"],
    "Calories": [360, 330, 155, 265, 579, 150, 208, 320, 402, 350],
    "Protein": [7.0, 31.0, 13.0, 9.0, 21.0, 10.0, 20.0, 4.0, 25.0, 25.0]
})

print(food)
print(food[food["Calories"] > 300])
print(food.groupby("Category")["Protein"].mean())
print(food.sort_values(by="Calories", ascending=False))
```

## Tapşırıq 4

```python
import pandas as pd

work = pd.DataFrame({
    "Name": ["Aysel", "Murad", "Nigar", "Aysel", "Kamran", "Murad", "Nigar", "Kamran"],
    "Project": ["Website", "App", "Website", "Database", "App", "Database", "App", "Website"],
    "Hours": [12, 15, 10, 8, 14, 9, 11, 7]
})

print(work)
print(work.groupby("Name")["Hours"].sum())
print(work.groupby("Project")["Hours"].sum())
hours_by_employee = work.groupby("Name")["Hours"].sum()
print("Ən çox saat işləyən əməkdaş:", hours_by_employee.idxmax())
print("İş saatı:", hours_by_employee.max())
```

## Tapşırıq 5

```python
import pandas as pd

tickets = pd.DataFrame({
    "Movie": ["Film A", "Film B", "Film C", "Film A", "Film B", "Film C", "Film A", "Film D", "Film B", "Film D", "Film C", "Film D"],
    "City": ["Bakı", "Gəncə", "Bakı", "Sumqayıt", "Bakı", "Gəncə", "Gəncə", "Bakı", "Sumqayıt", "Gəncə", "Sumqayıt", "Bakı"],
    "TicketsSold": [120, 90, 150, 80, 110, 75, 95, 130, 85, 100, 60, 140]
})

print(tickets)
print(tickets.groupby("Movie")["TicketsSold"].sum())
print(tickets.groupby("City")["TicketsSold"].sum())
tickets_by_movie = tickets.groupby("Movie")["TicketsSold"].sum()
print("Ən çox bilet satılan film:", tickets_by_movie.idxmax())
print("Satılan biletlərin sayı:", tickets_by_movie.max())
```
